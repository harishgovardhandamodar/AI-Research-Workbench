# Central Limit Theorem demo

Draw sample means from an exponential population and watch them become normal; measure how the standard error shrinks with sample size.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
rng = np.random.default_rng(1)
means = [rng.exponential(1.0, 30).mean() for _ in range(5000)]

fig, ax = plt.subplots(figsize=(5.5, 3.5))
ax.hist(means, bins=40, density=True, alpha=0.6, label='sample means')
xs = np.linspace(0.3, 1.7, 200)
ax.plot(xs, stats.norm.pdf(xs, 1.0, 1/np.sqrt(30)), color='#e05b5b', label='N(1, 1/sqrt(30))')
ax.set_xlabel('mean'); ax.set_ylabel('density'); ax.legend()
print(f'mean of means = {np.mean(means):.3f}, sd = {np.std(means):.3f} (theory 0.183)')
```

```python
sizes = np.array([2, 5, 10, 20, 50, 100])
stds = [np.std([rng.exponential(1.0, s).mean() for _ in range(2000)]) for s in sizes]

fig, ax = plt.subplots(figsize=(5.5, 3.5))
ax.plot(sizes, stds, 'o-', color='#4f8cff', label='empirical')
ax.plot(sizes, 1/np.sqrt(sizes), '--', color='#e05b5b', label='1/sqrt(n)')
ax.set_xlabel('sample size'); ax.set_ylabel('std of the mean')
ax.legend(); ax.set_title('Convergence of the standard error')
```
