# Logistic population growth

Compare growth curves for different intrinsic rates and inspect the growth-vs-density relationship.

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

def logistic(t, N, r, K): return r * N * (1 - N / K)
K = 1000.0
sol = solve_ivp(logistic, (0, 40), [10.0], args=(0.3, K), max_step=0.1)
sol_fast = solve_ivp(logistic, (0, 40), [10.0], args=(0.8, K), max_step=0.1)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(sol.t, sol.y[0], label='r = 0.3')
ax.plot(sol_fast.t, sol_fast.y[0], label='r = 0.8')
ax.axhline(K, color='#8b97a5', ls='--', label='K')
ax.set_xlabel('time'); ax.set_ylabel('N'); ax.legend()
ax.set_title('Logistic growth')
```

```python
N = np.linspace(0, 1.2*K, 200)
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(N, 0.5 * N * (1 - N / K), color='#4f8cff')
ax.axhline(0, color='#8b97a5', lw=0.8)
ax.set_xlabel('N'); ax.set_ylabel('dN/dt')
ax.set_title('Per-capita growth rate'); ax.set_xlim(0, 1.2*K)
```
