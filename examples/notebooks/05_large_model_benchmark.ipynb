# Model benchmark with cross-validation

Generate a synthetic classification problem, train three models (logistic regression, random forest, SVM) with 5-fold CV, and compare ROC curves.

```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import roc_curve, auc

X, y = make_classification(n_samples=800, n_features=20, n_informative=8,
                            n_redundant=4, random_state=3)
```

```python
models = {
    'LogisticRegression': LogisticRegression(max_iter=2000),
    'RandomForest': RandomForestClassifier(n_estimators=100, random_state=3),
    'SVM': SVC(probability=True, random_state=3),
}
cv = StratifiedKFold(5, shuffle=True, random_state=3)
for name, m in models.items():
    scores = cross_val_score(m, X, y, cv=cv, scoring='accuracy')
    print(f'{name:18s} acc = {scores.mean():.3f} ± {scores.std():.3f}')
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(6, 5))
for name, m in models.items():
    m.fit(X, y)
    fpr, tpr, _ = roc_curve(y, m.predict_proba(X)[:, 1])
    ax.plot(fpr, tpr, label=f'{name} (AUC={auc(fpr, tpr):.3f})')
ax.plot([0,1],[0,1], ls='--', color='#8b97a5')
ax.set_xlabel('false positive rate'); ax.set_ylabel('true positive rate')
ax.set_title('ROC curves'); ax.legend()
```
