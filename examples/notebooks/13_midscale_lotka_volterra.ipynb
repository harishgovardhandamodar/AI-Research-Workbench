# Lotka–Volterra predator–prey

Integrate the classic coupled ODEs and look at cycles in both time and phase space.

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

def lv(t, y, a, b, c, d):
    x, z = y
    return [a*x - b*x*z, c*x*z - d*z]
a, b, c, d = 1.1, 0.4, 0.1, 0.4
sol = solve_ivp(lv, (0, 60), [10, 5], args=(a, b, c, d), max_step=0.05)
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(sol.t, sol.y[0], label='prey')
ax.plot(sol.t, sol.y[1], label='predator')
ax.set_xlabel('time'); ax.legend(); ax.set_title('Populations over time')
```

```python
fig, ax = plt.subplots(figsize=(5.5, 5))
for ic in ([10, 5], [15, 8], [6, 12]):
    s = solve_ivp(lv, (0, 60), ic, args=(a, b, c, d), max_step=0.05)
    ax.plot(s.y[0], s.y[1], label=f'start {ic}')
ax.set_xlabel('prey'); ax.set_ylabel('predator')
ax.legend(); ax.set_title('Phase plane')
```
