# 1D heat diffusion

Solve the heat equation with an explicit finite-difference scheme: an initially hot spot spreading over time.

```python
import numpy as np
import matplotlib.pyplot as plt
alpha, L, nx = 0.01, 1.0, 50
dx = L / (nx - 1)
dt = 0.0002                    # dt <= dx^2 / (2 alpha) for stability
steps = 600
u = np.zeros(nx); u[nx//2 - 3:nx//2 + 3] = 1.0   # hot spot
history = [u.copy()]
c = alpha * dt / dx**2
for _ in range(steps):
    u = u + c * (np.roll(u, -1) + np.roll(u, 1) - 2*u)
    u[0] = u[-1] = 0
    if _ % 120 == 0: history.append(u.copy())
```

```python
x = np.linspace(0, L, nx)
fig, ax = plt.subplots(figsize=(6, 4))
for i, prof in enumerate(history):
    ax.plot(x, prof, label=f't = {i*steps//len(history)*dt:.3f}')
ax.set_xlabel('x'); ax.set_ylabel('temperature'); ax.legend()
ax.set_title('Temperature profiles')
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
im = ax.imshow(np.array(history), aspect='auto', origin='lower',
               extent=[0, L, 0, steps*dt], cmap='inferno')
ax.set_xlabel('x'); ax.set_ylabel('time');
fig.colorbar(im, label='temperature')
ax.set_title('Spacetime heat map')
```
