# Quick stats — comparing two groups

A tiny, fast experiment: summary statistics, a Welch t-test and an effect size for two simulated groups, with one histogram figure.

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt

rng = np.random.default_rng(0)
group_a = rng.normal(loc=100.0, scale=12.0, size=60)
group_b = rng.normal(loc=108.0, scale=14.0, size=55)
```

```python
print('group A: mean=%.2f sd=%.2f n=%d' % (group_a.mean(), group_a.std(ddof=1), len(group_a)))
print('group B: mean=%.2f sd=%.2f n=%d' % (group_b.mean(), group_b.std(ddof=1), len(group_b)))

t, p = stats.ttest_ind(group_a, group_b, equal_var=False)
pooled = np.sqrt((group_a.var(ddof=1) + group_b.var(ddof=1)) / 2)
d = (group_b.mean() - group_a.mean()) / pooled   # Cohen's d
print(f'Welch t = {t:.2f}, p = {p:.4g}, Cohen\'s d = {d:.2f}')
```

```python
fig, ax = plt.subplots(figsize=(5.5, 3.2))
ax.hist(group_a, bins=14, alpha=0.6, label='group A')
ax.hist(group_b, bins=14, alpha=0.6, label='group B')
ax.set_xlabel('measurement'); ax.set_ylabel('count')
ax.set_title('Two groups'); ax.legend()
```
