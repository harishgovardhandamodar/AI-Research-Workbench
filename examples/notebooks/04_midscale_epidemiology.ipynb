# SIR epidemic model

Integrate the classic susceptible–infected–recovered ODEs for different R0 values and compare outbreak size and peak timing.

```python
import numpy as np
from scipy.integrate import solve_ivp

def sir(t, y, beta, gamma):
    s, i, r = y
    return [-beta * s * i, beta * s * i - gamma * i, gamma * i]
```

```python
def run(R0, N=1_000_000, gamma=0.2, days=200):
    beta = R0 * gamma
    sol = solve_ivp(sir, (0, days), [0.999, 0.001, 0.0],
                    args=(beta, gamma), dense_output=False, max_step=1)
    s, i, r = sol.y
    peak = int(np.argmax(i))
    attack = 1 - s[-1]
    return sol.t, i, r, peak, attack

results = {}
for R0 in (1.5, 2.5, 4.0):
    results[R0] = run(R0)
    t, i, r, peak, attack = results[R0]
    print(f'R0={R0}: peak infected {i[peak]/1e3:.1f}k at day {peak}, '
          f'attack rate {attack*100:.1f}%')
```

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(7, 4))
for R0, (t, i, r, peak, attack) in results.items():
    ax.plot(t, i / 1e3, label=f'R0 = {R0}')
ax.set_xlabel('days'); ax.set_ylabel('infected (thousands)')
ax.set_title('SIR outbreak — infected over time'); ax.legend()
```
