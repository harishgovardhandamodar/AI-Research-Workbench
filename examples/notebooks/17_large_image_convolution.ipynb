# Image filtering: blur & edge detection

Build a synthetic image, apply a Gaussian blur and Sobel edge detection with scipy.ndimage.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import ndimage
rng = np.random.default_rng(4)
img = rng.normal(0, 0.1, (128, 128))
img[30:50, 40:80] += 1.5      # bright rectangle
img[70:90, 30:60] -= 1.0      # dark block
yy, xx = np.ogrid[:128, :128]
img += 0.8 * np.exp(-((xx-100)**2 + (yy-20)**2) / 200)
img += rng.normal(0, 0.15, img.shape)
```

```python
blur = ndimage.gaussian_filter(img, sigma=2.0)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, im, t in zip(axes, [img, blur, img - blur],
                      ['original', 'Gaussian blur', 'difference']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
```

```python
gx = ndimage.sobel(blur, axis=0); gy = ndimage.sobel(blur, axis=1)
edges = np.hypot(gx, gy)
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
for ax, im, t in zip(axes, [edges, edges > 0.4, edges > 0.8],
                      ['Sobel magnitude', 'threshold 0.4', 'threshold 0.8']):
    ax.imshow(im, cmap='gray'); ax.set_title(t); ax.axis('off')
```
