# Synthetic data from the privacy MCP tools

Generate a schema-preserving synthetic version of the clinical cohort, then compare utility (means, spreads, distributions) against the real data before sharing.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd()))   # repo root (kernel cwd)

from mcp_servers import privacy_tools as pt

# Ensure the cohort CSV exists (generated by 20_privacy_assessment).
if not Path('examples/privacy/clinical_cohort.csv').exists():
    from examples.privacy.clinical_cohort import build_cohort
    build_cohort(200, seed=7).to_csv('examples/privacy/clinical_cohort.csv', index=False)
print(pt.generate_synthetic_tabular(
    'examples/privacy/clinical_cohort.csv', num_rows=1000, method='smoothed', seed=7))
```

```python
print(pt.synthetic_data_quality_report(
    'examples/privacy/clinical_cohort.csv',
    'examples/privacy/synthetic_clinical_cohort_1000.csv'))
```

```python
import pandas as pd
import matplotlib.pyplot as plt
real = pd.read_csv('examples/privacy/clinical_cohort.csv')
synth = pd.read_csv('examples/privacy/synthetic_clinical_cohort_1000.csv')
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
for ax, col in zip(axes, ['age', 'visit_amount_usd']):
    ax.hist(real[col], bins=25, alpha=0.6, label='real', color='#4f8cff')
    ax.hist(synth[col], bins=25, alpha=0.6, label='synthetic', color='#35c4b6')
    ax.set_xlabel(col); ax.set_ylabel('count'); ax.legend()
axes[0].set_title('Age — real vs synthetic')
axes[1].set_title('Visit amount — real vs synthetic')
```
