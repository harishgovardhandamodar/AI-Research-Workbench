# Double pendulum

Integrate the chaotic double pendulum and check energy conservation while watching the trajectory.

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
g, m, L = 9.81, 1.0, 1.0

def dp(t, y):
    t1, w1, t2, w2 = y
    d = t1 - t2
    a = (m*L*L); b = m*L*L*np.cos(d)
    c = m*L*L; f = m*L*L*np.cos(d)
    e = -m*g*L*2*np.sin(t1) - m*g*L*np.sin(t2)
    g0 = m*L*L*(w1**2*np.sin(d) + w2**2*np.sin(d)) - m*g*L*np.sin(t2)
    return [w1, w2,
            (e*c - b*g0) / (a*c - b*b),
            (a*g0 - e*f) / (a*c - b*b)]
sol = solve_ivp(dp, (0, 30), [np.pi/2, 0, np.pi/2, 0], max_step=0.02)
```

```python
t1, w1, t2, w2 = sol.y
x2 = L*np.sin(t1) + L*np.sin(t2)
y2 = -L*np.cos(t1) - L*np.cos(t2)
fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(x2, y2, lw=0.8, color='#4f8cff')
ax.set_aspect('equal'); ax.set_title('Trajectory of the second bob')
```

```python
energy = (0.5*(w1**2 + 2*w1*w2*np.cos(t1-t2) + w2**2)
         - g/L*(2*np.cos(t1) + np.cos(t2)))   # per unit m, L
fig, axes = plt.subplots(1, 2, figsize=(10, 3.8))
axes[0].plot(sol.t, energy, color='#35c4b6')
axes[0].set_xlabel('t'); axes[0].set_ylabel('E / mgL')
axes[0].set_title(f'Energy drift {100*abs(energy.max()-energy.min())/abs(energy.mean()):.3f}%')
axes[1].scatter(t1, t2, s=1, color='#d9a441')
axes[1].set_xlabel('theta1'); axes[1].set_ylabel('theta2')
axes[1].set_title('theta1 vs theta2')
```
