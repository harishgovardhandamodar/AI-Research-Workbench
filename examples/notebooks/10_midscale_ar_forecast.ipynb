# AR(2) time series & forecasting

Simulate an autoregressive series, inspect its autocorrelation structure and forecast ahead.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(9)
n = 300; phi1, phi2 = 0.6, -0.25
x = np.zeros(n)
for t in range(2, n):
    x[t] = phi1*x[t-1] + phi2*x[t-2] + rng.normal(0, 1)
print(f'variance = {x.var():.3f} (theory ~1.4)')
```

```python
def acf(x, lag):
    x = x - x.mean()
    c0 = x @ x
    return np.array([(x[:-k] @ x[k:]) / c0 for k in range(1, lag+1)])
ac = acf(x, 20)
fig, axes = plt.subplots(1, 2, figsize=(10, 3.5))
axes[0].stem(range(1, 21), ac)
axes[0].axhline(1.96/np.sqrt(n), color='#e05b5b', ls=':'); axes[0].axhline(-1.96/np.sqrt(n), color='#e05b5b', ls=':')
axes[0].set_title('ACF')
axes[1].plot(x); axes[1].set_title('Series')
axes[1].set_xlabel('t')
```

```python
p = 10
Xm = np.column_stack([x[p-1-i : n-1-i] for i in range(p)])
pac = np.linalg.lstsq(Xm, x[p:], rcond=None)[0]

fig, ax = plt.subplots(figsize=(6, 3.5))
ax.stem(range(1, p+1), pac)
ax.set_title('PACF (regression-based)'); ax.set_xlabel('lag')

# 1-step-ahead recursive forecast
steps = 30
f = list(x[-2:])
for _ in range(steps):
    f.append(phi1*f[-1] + phi2*f[-2])
fig2, ax2 = plt.subplots(figsize=(7, 3.5))
ax2.plot(range(n), x, color='#4f8cff', label='observed')
ax2.plot(range(n, n+steps), f[2:], color='#e05b5b', ls='--', label='forecast')
ax2.set_xlabel('t'); ax2.legend(); ax2.set_title('AR(2) forecast')
```
