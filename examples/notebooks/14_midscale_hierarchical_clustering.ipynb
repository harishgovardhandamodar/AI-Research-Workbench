# Hierarchical clustering of expression profiles

Build synthetic samples across features, cluster them hierarchically and inspect the resulting tree + heatmap.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from scipy.spatial.distance import pdist
rng = np.random.default_rng(17)
n_samples, n_feat = 40, 25
types = np.repeat([0, 1, 2], [15, 12, 13])
X = rng.normal(0, 1, (n_samples, n_feat))
for t in range(3):
    X[types == t] += rng.uniform(0.8, 2.0, n_feat)
```

```python
Z = linkage(pdist(X), method='ward')
fig, ax = plt.subplots(figsize=(7, 3.5))
dendrogram(Z, ax=ax, color_threshold=8, no_labels=True)
ax.set_title('Ward dendrogram')
```

```python
order = dendrogram(Z, no_plot=True)['leaves']
fig, ax = plt.subplots(figsize=(4.5, 6))
im = ax.imshow(X[order].T, aspect='auto', cmap='coolwarm', vmin=-2, vmax=3)
fig.colorbar(im, label='expression')
ax.set_xlabel('sample'); ax.set_ylabel('feature')
ax.set_title('Heatmap (dendrogram order)')
```
