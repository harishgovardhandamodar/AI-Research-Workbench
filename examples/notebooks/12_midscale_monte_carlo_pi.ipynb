# Monte Carlo estimation of pi

Throw random points into the unit square and see the estimate converge as the number of samples grows.

```python
import numpy as np
import matplotlib.pyplot as plt
rng = np.random.default_rng(2)
n = 4000
pts = rng.uniform(0, 1, (n, 2))
inside = (pts[:, 0]**2 + pts[:, 1]**2) <= 1
fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(pts[inside, 0], pts[inside, 1], s=3, color='#4f8cff')
ax.scatter(pts[~inside, 0], pts[~inside, 1], s=3, color='#e05b5b')
ax.set_aspect('equal'); ax.set_title(f'n = {n}, pi ~ {4*inside.mean():.4f}')
print(f'estimate after {n} samples: {4*inside.mean():.4f}')
```

```python
cum = 4 * np.cumsum(inside) / np.arange(1, n+1)
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.plot(cum, color='#35c4b6', lw=1)
ax.axhline(np.pi, color='#e05b5b', ls='--', label='pi')
ax.set_xlabel('samples'); ax.set_ylabel('estimate')
ax.legend(); ax.set_title('Convergence of the estimate')
```
